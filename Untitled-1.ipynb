```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img=cv.imread(r'C:\Users\kavin\Downloads\runway.png',cv.IMREAD_GRAYSCALE)
img1=img/255

r1, r2 = 0.2, 0.8
output = np.zeros_like(img1)

output[img1 >= r1] = 0
output[(img1 >= r1) & (img1 <= r2)] = \
    (img1[(img1 >= r1) & (img1 <= r2)]-r1)/(r2 - r1)
output[img1 > r2] = 1

output1 = np.uint8(output * 255)

cv.imshow('Resulting Image', output1)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img=cv.imread(r'C:\Users\kavin\Downloads\highlights_and_shadows.jpg')
lab = cv.cvtColor(img,cv.COLOR_BGR2LAB)
L, a, b = cv.split(lab)

gamma = 0.6
L_norm = L/255
L_gamma = np.power(L_norm, gamma)
L_gamma = np.uint8(L_gamma * 255)

lab_out = cv.merge((L_gamma, a, b))
out = cv.cvtColor(lab_out, cv.COLOR_LAB2BGR)

cv.imshow('Output Image', out)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img=cv.imread(r'C:\Users\kavin\Downloads\runway.png',cv.IMREAD_GRAYSCALE)

hist,_=np.histogram(img.flatten(),256,[0,256])
cdf=hist.cumsum()
cdf=cdf/cdf[-1]
out=np.interp(img.flatten(),np.arange(0,256),cdf*255)
out=out.reshape(img.shape).astype(np.uint8)

plt.imshow(out,cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')
plt.show()
```
